# 📊 QSAFE Nepal - Hybrid NLU & RAG Benchmark Evaluation
This Jupyter Notebook executes the modular evaluation framework on `eval_dataset.json`.

### Metrics Computed:
- **NLU Intent Classification**: Accuracy, Precision, Recall, Macro F1, Script Breakdown (`en`, `np_devanagari`, `np_romanized`)
- **ChromaDB Vector Search**: Hit Rate@3, Mean Reciprocal Rank (MRR), Cosine Distance
- **LLM Generation**: ROUGE-L F1, Semantic Cosine Similarity
- **System Latencies**: Intent latency, Retrieval latency, Generation latency
- **Failure Analysis**: Categorized routing/retrieval/generation failures saved to `evaluation_failures.json`

```python
import json
import pandas as pd
from evaluation_framework import HybridPipelineEvaluator

# Load evaluation dataset
with open('eval_dataset.json', 'r', encoding='utf-8') as f:
    eval_dataset = json.load(f)

print(f"✅ Successfully loaded {len(eval_dataset)} evaluation samples from eval_dataset.json")
```

```python
# Execute Hybrid Pipeline Benchmark
evaluator = HybridPipelineEvaluator()
summary_df, script_df, failures_df = evaluator.evaluate_dataset(eval_dataset, top_k=3, generation_threshold=0.5)
```
